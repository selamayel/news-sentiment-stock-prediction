```python
## 1. Descriptive Statistics

import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

df = pd.read_csv("../yfinance_data/MSFT_historical_data.csv")  

# Convert Date column to datetime
df['Date'] = pd.to_datetime(df['Date'])

# Plot distributions of numerical columns
sns.histplot(df['Close'], bins=30, kde=True)
plt.title('Distribution of Closing Prices')
plt.xlabel('Close')
plt.show()

sns.histplot(df['Volume'], bins=30, kde=True)
plt.title('Distribution of Volume')
plt.xlabel('Volume')
plt.show()

# Group by date and count entries (should be 1 per date for stock data)
daily_counts = df.groupby(df['Date'].dt.date).size()

# Plot
daily_counts.plot(figsize=(12,5))
plt.title('Entry Count Over Time')
plt.xlabel('Date')
plt.ylabel('Number of Entries')
plt.grid(True)
plt.show()
```
